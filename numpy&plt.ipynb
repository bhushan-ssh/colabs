```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.linspace(0,1,10)
y=x
plt.plot(x, y)
plt.scatter(x, y)
plt.title("Linear Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid()
```
