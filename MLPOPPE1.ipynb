```python
def category(PRICE):
  if PRICE<10:
    return "Category 1"
  if 20>PRICE>=10:
    return "Category 2"
  if 30>PRICE>=20:
    return "Category 3"

  if 40>PRICE>=30:
    return "Category 4"
  else:
    return "Category 5"
df["CATEGORY"]=df['PRICE'].apply(category)
df.head()
df['CATEGORY'].value_counts()
```
